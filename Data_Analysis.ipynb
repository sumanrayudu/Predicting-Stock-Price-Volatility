```python
import yfinance as yf

tickers = ["AAPL","MSFT","GOOGL","AMZN","NVDA","TSLA","META"]
df = yf.download(tickers, start="2015-01-01", end="2024-12-31", group_by="ticker")

```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt

nv = df["NVDA"]["Close"].reset_index()
plt.figure(figsize=(10,4))
plt.plot(nv["Date"], nv["Close"])
plt.xlabel("Date")
plt.ylabel("NVDA Close Price")
plt.title("Nvidia Closing Price (2015–2024)")
plt.show()
```

```python
nv["Return"] = nv["Close"].pct_change()
nv["Vol30"] = nv["Return"].rolling(window=30).std() * (252**0.5)
plt.figure(figsize=(10,4))
plt.plot(nv["Date"], nv["Vol30"])
plt.xlabel("Date")
plt.ylabel("Annualized 30-day Volatility")
plt.title("Nvidia Rolling Volatility")
plt.show()
```
